## String:
characters written inside double or single quotes are refered as string in python

```python
name = "Muhammad Umair"
subject = 'Physics'
print(name,type(name))
print(subject,type(subject))
```

```python
# Length of string can be determined by using len method. 
# syntax: len(name of variable containg string data type)
print(len(name))
print("Length of the string",subject,"is",len(subject))
```

```python
# Escape character \n is used to write mutipline line string in python.
intro = " My name is Muhammad Umair. \n I teach Physics."
print(intro)
```

```python
# \t can be used to give a space equal to 1 tab or 5 spacebars.
new = " I live in \t Abbottabad, Pakistan."
print(new)
```

### Strings Indexing:
Indexing of strings refers to the position of each individual character ( including spaces ) in the strig. Indexing starts form 0 in going from left to right of string and it starts form -1 in going from right to left of string.

```python
name = "Muhammad Umair"
l1 = name[3]
l2 = name[5]
l3 = name[-2]
l4 = name[-1]
print(l1,l2,l3,l4)
```

##### Range of indexing

```python
bro = name[2:-5]
print(bro)
```

##### Stepwise indexing

```python
new = name[0:9:2]
print(new)
```
