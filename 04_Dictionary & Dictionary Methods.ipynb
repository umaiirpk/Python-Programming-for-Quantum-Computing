## Dictionary Data Type:
Dictionary is an indexed but unsequenced collection of various data type members written inside curly brackets. The member of Dictionary is a key-value pair, a key is seperated by value using colon ":" . Both the key and value can store any data type. Dictionary is muteable i.e. we can add a new key-value pair into it and also we can remove key-value pair already present in it therefore it doesn't have fixed length.

```python
my_dictionary = {"name":"Muhammad Umair", "Roll no":207811, "cgpa":3.73,"university":"AUST"}
print(my_dictionary)
print(type(my_dictionary))
```

### Nested Dictionary

```python
# We can also place a dictionary as value of another dictionary
m2 = {"teacher":"Muhammad Umair", "classes":7,"subjects":{"s1":"Phy","s2":"Math","s3":"Che"}}
print(m2)
print(type(m2))
```

```python
# Note: Dictionary doesn't allow duplicate key-value pairs rahter they are stored only ones.
m3 = {1:2,3.4:3,"number":9,"ndict" : {1:3,3:5,5:4}, "number":9, 3.4:3,1:2}
print(m3)
print(type(m3))
```

### Dictionary Methods:

```python
# Auccessing individual values of different keys in a dictionary
m2 = {"teacher":"Muhammad Umair", "classes":7,"subjects":{"s1":"Phy","s2":"Math","s3":"Che"}}
n = m2["teacher"]
print(n)
m = m2["subjects"]["s1"]
print(m)
```

```python
m2.keys() # will give all keys of the mentioned dictionary
```

```python
m2.values()# will give all values of the mentioned dictionary
```

```python
# Adding a new key-value pair in a dictionary
m2["year"]=2026
print(m2)
```

```python
m2["classes"] = 8
print(m2)
```
