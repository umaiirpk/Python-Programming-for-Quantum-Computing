## Set Data Type:
A set is an unsequenced and unindexed collection of distinct data type elements written inside curly brackets. Elements of the set can be of any data type (int, float, str, bool). Sets are muteable i.e. we can add or remove elements in a set. Set doesn't allow duplicate elements, it only store them ones. Various arthemic set operations i.e. union, intersection etc can be performed on python sets.

```python
s1 = {"a","e","i","o","u","a",4,3.73}
print(s1)
print(type(s1))
print("notice by printing s1 set again the sequence of elements has changed",s1)
```

```python
# Writting an empty set in python
ems= set()
print(type(ems))
```

### Set Methods:

```python
# Removing an element of set
s1 = {"a","e","i","o","u","a",4,3.73}
s1.remove(3.73)
print(s1)
```

```python
# Adding a new element to a set
s1.add("y") # note add method for set only take single argument
print(s1)
```

```python
# Union of Two Sets
s2 = {1,2,3,4,5,6,7,8,9}
s3 = s1.union(s2)
print("The union of set s1 and s2 is", s3)
```

```python
# Intersection of two sets
s4 = s2.intersection(s1)
print("Intersection of set s1 and set s2 is", s4)
```
