### Tuple Data type:
Tuple is the collection of sequenced, fixed, and indexed elements written inside round brackets. Elements of Tuple can be of any data type (int, float, str, bool). Tuple is immutable i.e. elements can not be changed and no element can be added or removed from a Tuple. Hence Tuple has a fixed length.Tuple allow duplicate elements.

```python
t = (1,2,3,1,1,3,2,3,"a","b","c",[5,6],{8,9,0},{3:4,6:7,1:2})
print(t, type(t))
```

```python
t.index(3) = 9 # will give error because tuple is immutable.
```

### Tuple Methods:

```python
t.count(3) # count method gives the frequency of occurance of an element of tuple.
```

```python
t.index(3) # index method give the first occurance index of an element.
```
