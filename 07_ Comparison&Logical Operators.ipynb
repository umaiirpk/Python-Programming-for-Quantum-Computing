## Comparison Operators:

```python
# Equal to operator: "=="
5==5
4==8
```

```python
# Not equal to " != "
4 != 8
```

```python
# greater than " > "
4>8
```

```python
# less than: " < "
4 < 8
```

```python
# Greater than or equal to : " >="
4>=8
```

```python
# Less than or equal to : " <= " 
4<=8
```

## Logical Operators:

```python
# and operator: for and operator all statements must be true 
4<8 and 7<9
```

```python
# or operator: for or operator any one of the statements must be true
4>5 or 5>7
```

```python
not 4==5 # nor operator reverses the statements i.e. turns true into fales and fales into true
```
