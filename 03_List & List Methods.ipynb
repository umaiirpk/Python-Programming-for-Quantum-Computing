## List Data Type:
List is indexed sequence of different data types written in side square brackets[]. List is a muteable i.e. we can add new member to it or can remove existing member from it; therefore, lists don't have a fixed length. We can have duplicate members inside the same list. 

```python
vol = ["a","e","i","O","u"]
num = [1,2,3,4,5,5,2]
new = [1,2,3,4.4,5.8,"j","r"]
print(type(vol))
print(type(num))
print(type(new))

```

```python
# List Indexing
v = vol[2]
n = num[-4]
w = new[-3]
print(v,n,2)
```

```python
# List concatination
y = num + new
print(y)
print(type(y))
```

### List Methods:

```python
y.append(9) # append will only take one argument hence it will only add only one new member in the end of list.
print(y)
```

```python
y.insert(12,13)
print(y)
```

### Nested Lists

```python
r = [1,2,3.3,[3,4,5]]
new_list = r[3]
print(new_list)
print(type(new_list))
```
