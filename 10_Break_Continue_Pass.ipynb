### Break:

```python
my_list = [10,20,30,40,50,60]
for number in my_list:
    if number == 50:
        break
    print(number**2)
    
```

### Continue:

```python
my_tuple = (10,20,30,40,50)
for numb in my_tuple:
    if numb == 30:      # it will skip 30
        continue
    print(numb*2)
```
