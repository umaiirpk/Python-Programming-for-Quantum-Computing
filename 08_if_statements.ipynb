## if statement:
if statementis is used to check whether given variable hold some condition true. 
if statement is written with "if" keyward at the start. If the variable don't hold a certain condition true the it doesn't print any thing.

```python
x = int(input("Write the first number of you chooise"))
y = int(input("write the second number of your chooise"))
if x > y:
    print("Output: x is greater then y")
```

## elif statement: 
elif statement is used to check whether the given variable hold true some other requried conditions than it will give the conditioned statement as output. elif statement starts with "elif" keyward. Syntax: elif (requried condition): 

```python
x = int(input("Write the first number of you chooise"))
y = int(input("write the second number of your chooise"))
if x > y:
    print("Output: x is greater then y")
elif x < y:
    print("Output: x is less then y")
```

## else statement: 
else statement is when all all conditions are not true. it start with only "else" keyword with colon.

```python
x = int(input("Write the first number of you chooise"))
y = int(input("write the second number of your chooise"))
if x > y:
    print("Output: x is greater then y")
elif x < y:
    print("Output: x is less then y")
else:
    print("Output: x is equal to y")
```

### Using Logical operators in if statements: 

```python
x = int(input("Enter your favorite number"))
if x%4 == 0 or x%6 == 0 or x%8 == 0:
    print("x is divisible by 2")
elif x%2 == 0:
    print("x is an even number")
else:
    print("need more information about x")
```

### Nested if statements:

```python
y = int(input("enter a number"))
if y>10:
    print("nothing")
if y>20:
    print("something")
elif y==50:
    print("half century")
else:
    print("need more information")
```
